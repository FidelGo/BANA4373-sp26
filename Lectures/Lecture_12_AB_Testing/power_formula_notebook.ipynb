# Power Formula Simulation


This notebook helps you understand how sample size depends on:
- Effect size (δ)
- Variance (σ²)
- Power
- Significance level


```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python

def required_n(alpha=0.05, power=0.8, sigma=1, delta=0.5):
    z_alpha = norm.ppf(1 - alpha/2)
    z_beta = norm.ppf(power)
    n = ((z_alpha + z_beta)**2 * 2 * sigma**2) / delta**2
    return n

required_n()
```

## How sample size changes with effect size

```python

deltas = np.linspace(0.1, 1.0, 50)
ns = [required_n(delta=d) for d in deltas]

plt.plot(deltas, ns)
plt.xlabel("Effect size (delta)")
plt.ylabel("Required sample size per group")
plt.title("Sample size vs Effect size")
plt.show()
```

## Key takeaway: smaller effects require much larger samples
